```python
# Qn3) output- dict= {'a':['apple','ate',"b":[.......],.............]
list =['apple','bat','cat','ball','cow','ate']
l1=[list[i] for i in range(len(list)) if list[i][0]=="a"]
l2=[list[i] for i in range(len(list)) if list[i][0]=="b"]
l3=[list[i] for i in range(len(list)) if list[i][0]=="c"]
dict2={}
for i in range(len(list)):
    if list[i][0]=="a":
        dict2[list[i][0]]=l1
    elif list[i][0]=="b":
        dict2[list[i][0]]=l2
    else:
        dict2[list[i][0]]=l3
dict2
```

```python
# Qn4) d={1:'a',2:'b',3:'c'}   interchange

dict={'a':1,'b':2,'c':3}
lt1=[i for i in dict]
lt2=[dict[i] for i in dict]
d1={}
for i,j in zip(lt1,lt2):
    d1[j]=i
d1
```

```python
 # Qn5) common = {'b':20}  Common key:value pair
d1={'a':1,'b':20,'c':30}
d2={'a':11,'b':20,'c':50}
d4={}
for i,j in zip(d1,d2):
    if i==j and d1[i]==d2[j]:
        d4[i]=d1[j]
d4
```
